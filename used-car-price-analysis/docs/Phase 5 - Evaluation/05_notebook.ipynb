### Phase 5 (Evaluation) - Used Car Price Prediction

### 1. Evaluate the Obtained Results

#### 1.1 Overview

In this phase, we evaluate the results obtained from Phase 4 in greater depth. The objective is not only to compare the numerical performance of the model but also to understand why certain models performed better and what this means for the project moving forward. 

The evaluation includes both **quantitative metrics** (R², MAE, RMSE) and **qualitative observations**, such as prediction stability across price ranges, residual distribution patterns, model biases, and the interpretability of feature importance.

This stage ensures that the entire modeling pipeline was executed correctly, identifies potential weaknesses in model behavior, and guides decisions regarding tuning, feature engineering, and future improvements. The focus is placed on **accuracy, robustness, generalization ability,** and **practical usability** for real-world used car price prediction.


#### 1.2 Evaluation of Target Choice: Price vs Log(Price)

Before comparing individual models, we first evaluated the difference between predicting **raw car prices** and predicting **log-transformed prices**.

##### **Key observations:**

When models were trained on **raw prices**, they performed reasonably well for low- and mid-priced vehicles. However, prediction errors became much larger for expensive cars. These large errors had a strong negative impact on MAE and RMSE, even when the model performed well for most vehicles.

Visual inspection of predictions and residuals showed that **errors increased as car prices increased**. This means the models were less reliable for high-priced vehicles compared to cheaper ones.

After applying a **log transformation** to the target, the price distribution became more balanced. This helped models learn more consistently across all price ranges. The influence of extremely expensive vehicles was reduced, and overall prediction stability improved.

As a result, **R² scores increased after log transformation**, especially for tree-based ensemble models such as Random Forest, LightGBM, and CatBoost.

##### **Conclusion:**

Predicting **log(price)** creates a more balanced learning problem. It improves model performance across the full price range and leads to lower MAE and RMSE when predictions are converted back to original price values. For this reason, log(price) was selected as the main target for final model evaluation.



#### 1.3 Summary of Model Performance

The top models were evaluated using both log-transformed targets (for R²) and original price units (for MAE/RMSE). Key findings include:

* **Random Forest (tuned**) delivered the strongest performance with
  **R² = 0.946, MAE = 2,441,** and **RMSE = 5,823**, showing excellent predictive accuracy and robust behavior across the full price distribution.

* **LightGBM (tuned)** and CatBoost (tuned) were close competitors, performing slightly worse in MAE/RMSE but maintaining high R² and stable predictions.

* **XGBoost (untuned)** performed surprisingly well without tuning, demonstrating strong default parameter robustness.

* **MLP Regressor (tuned)** achieved reasonable R² on the log scale (0.887) but produced significantly higher errors in raw-price units, struggling especially with high-priced outliers.

Residual analysis shows that tree-based models produce tighter, symmetric residual distributions around zero, indicating unbiased performance. In contrast, neural networks display wider residual spread, particularly for expensive vehicles.

### 2. Analyze and Compare the Results of Individual Models

#### 2.1 Tree-Based Ensembles (Random Forest, LightGBM, CatBoost, XGBoost)

##### Strengths

* Capture non-linear patterns effectively

* Handle complex feature interactions

* Offer strong performance even with minimal tuning (especially XGBoost and Random Forest)

* Produce stable predictions with narrow residual distributions

##### Weaknesses

* Training time grows with dataset size

* Hyperparameter tuning for boosting algorithms can be computationally expensive

#### 2.2 Neural Networks (MLP)

##### Strengths

* Highly flexible architecture capable of modeling subtle relationships

* Can generalize well when enough training data and tuning are available

##### Weaknesses

* Very sensitive to feature scaling and regularization

* Requires more training data than tree models

* Underperforms on raw-price metrics, especially for luxury or uncommon vehicles

#### 2.3 Linear Models (Linear Regression, Ridge)

##### Strengths

* Simple and interpretable

* Very fast to train and straightforward to implement

##### Weaknesses

* Fail to capture non-linear relationships

* Extremely high errors on expensive cars

* Not suitable for datasets with complex interactions, like used car pricing

#### 2.4 Feature Importance Comparison

Across all models—tree ensembles and neural networks alike—the most influential predictors are consistently:

- **Horsepower and torque (engine performance)**

- **Year (vehicle age)**

- **Mileage (usage)**

Neural networks place slightly more emphasis on **horsepower**, while tree models balance horsepower, year, and mileage more evenly.

This confirms domain knowledge that vehicle performance, age, and wear are the dominant drivers of used car price.

### Residual Analysis

* For the top tree models, residual distributions are tightly **centered around zero**, indicating **unbiased prediction** and **consistent performance.**

* The MLP Regressor shows **wider residual spread**, particularly for high-priced vehicles, reflecting its higher MAE/RMSE.

* All models show some large-error cases for rare luxury vehicles, suggesting a need for:

    - additional high-end car features, or
    - model segmentation (separate luxury vs. mainstream models).

### Feature Importance Insights

Feature importance results across models provide strong validation for model correctness:

* Horsepower and torque consistently rank as top predictors.

* Year and mileage remain key indicators of car value depreciation.

* Secondary features such as size, body type, fuel type, and seating capacity contribute but have smaller effects.

The agreement across methods supports the conclusion that the models capture real underlying price relationships in the dataset.

### 3. Review Process

To validate the modeling workflow, each stage of the pipeline was reviewed for correctness and consistency:

* Preprocessing steps—including scaling and log transformation—were applied uniformly.

* Predictions were evaluated both in **log-transformed** and **original price units** to ensure meaningful interpretation.

* Residuals and error patterns were examined to identify over- or underestimation trends.

* Feature importance was inspected to confirm that the models rely on logically relevant predictors (e.g., mileage, year, horsepower).


The review confirmed that the models were implemented correctly. Log-transforming the target variable significantly improved prediction stability and reduced the influence of high-value outliers, especially for expensive vehicles.

### 4. Decide the Next Phases of the Project

**Recommendations for further development:**

**1. Model Optimization:**

* Extend hyperparameter tuning for XGBoost, CatBoost, and neural networks.

* Explore ensembles combining tree models and neural networks.

**2. Feature Engineering / Data Enhancement:**

* Incorporate additional data such as:

  - service/maintenance history
  - accident records
  - trim-level features
  - regional demand

* Handle rare luxury vehicles separately to reduce error spikes.

**3. Training Improvements:**

* Increase MLP training epochs, add dropout/batch normalization.

* Explore deeper architectures and alternative activation functions.

* Add polynomial or interaction terms for linear models to capture relationships.

**4. Evaluation Enhancements:**

* Use **k-fold cross-validation** for more reliable performance estimates.

* Introduce **weighted error metrics** if business priorities emphasize luxury or premium vehicles.

* Analyze per-price-segment performance (budget cars vs high-end cars).

### 5. Additional Questions
<br>

**How did machine and/or deep learning with different methods work?**

* **Tree-based ensemble models** (Random Forest, LightGBM, CatBoost, XGBoost) performed the best overall, capturing non-linear patterns and strong feature interactions.

* **Neural networks (MLP)** showed moderate performance, benefiting from the log-transformed target but still struggling with rare, high-priced cases.

* **Linear models** consistently underperformed because they cannot model complex relationships present in real-world car pricing.

**How well do different methods work?**

* **Tree-based ensembles** achieved top accuracy, low errors, and stable residuals, so it's ideal for practical deployment.

* **Neural networks** performed moderately well on log-transformed targets but struggled with raw-price accuracy.

* **Linear models** were unable to describe real pricing variation and produced high MAE and RMSE.

**What needs further development in the models?**

* Tuning MLP and XGBoost for improved performance.

* Additional features that capture car condition, history, and market-specific variables.

* Techniques for reducing errors on extremely high-priced vehicles.

* Exploration of ensemble or stacking models.

**How could you further optimize the model?**

* Conduct larger-scale hyperparameter optimization (Grid, Random, Bayesian).

* Build engineered features (age × mileage, engine size ratio, depreciation rate).

* Use specialized loss functions (Huber loss, quantile loss) to handle outliers.

* Apply weighted metrics to prioritize business-critical price ranges.

**Other considerations about the task:**

* Used car prices are highly skewed; log transformation is essential.

* Residuals indicate most errors come from **rare high-priced vehicles** (luxury vehicles).

* Tree ensembles provide the most interpretable and robust models for business use.
